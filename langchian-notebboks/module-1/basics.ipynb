## Check some basics

```python
import os
import json
from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
from langchain_ollama import ChatOllama

qwen = ChatOllama(model="qwen2.5:7b", temperature=0.5)
```

```python
from langchain_core.messages import HumanMessage

# create a message


msg = HumanMessage(content="Hello, how are you?", name="Bilalox")

# Message List

messages = [msg]

# Invoke the model with list of messages

results = qwen.invoke(messages)
        
        
```

```python
results
```

```python
print(results.content)
print(results.usage_metadata.get('input_tokens'))
print(results.usage_metadata.get('output_tokens'))
print(results.usage_metadata.get('total_tokens'))
```

We can say that we get AIMessage Answer.  We can also call the model with Strings

```python
qwen.invoke("Hello, how are you?")
```
